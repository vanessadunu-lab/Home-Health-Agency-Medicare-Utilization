```python
base_url = "https://data.cms.gov/data-api/v1/dataset/6c63099b-0794-40a0-925c-51a66b9b9901/data"
```

## Sending GET Request

```python
import requests
```

```python
all_data = []
size = 5000
offset = 0

while True:
    params = {
        "size" : size,
        "offset": offset
    }
    response = requests.get(base_url, params=params)
    data=response.json()

    if not data:
        break

    all_data.extend(data)
    offset += size

import pandas as pd
home_health_raw=pd.DataFrame(all_data)
```
